## Limpieza de Datos
Conversion del date_added a datatime

```python
df_movies['date_added']=pd.to_datetime(df_movies['date_added'],errors='coerce')
df_shows['date_added']=pd.to_datetime(df_shows['date_added'],errors='coerce')
```

Reemplazo de valores 0 de rating/budget/revenue por NaN (son datos faltantes)

```python
df_movies.loc[df_movies['rating']== 0, 'rating']=np.nan
df_movies.loc[df_movies['budget']== 0, 'budget']=np.nan
df_movies.loc[df_movies['revenue']== 0, 'revenue']=np.nan
```

Eliminamos filas sin títulos

```python
df_movies=df_movies.dropna(subset=['title'])
df_shows=df_shows.dropna(subset=['title'])
```

Eliminar duplicados

```python
df_movies=df_movies.drop_duplicates(subset=['show_id'])
df_shows=df_shows.drop_duplicates(subset=['show_id'])
```

```python
### Se preparan los géneros: separarlos múltiples géneros y explodirlo
df_movies_genres=df_movies[df_movies['genres'].notna()].copy()
df_movies_genres['genres']=df_movies_genres['genres'].str.split(', ')
df_movies_genres=df_movies.explode('genres')
print(f'Géneros preparados:\n\n {df_movies_genres['genres'].unique()}')
```

```python
top_genres=df_movies_genres['genres'].value_counts().head(10).index
df_top_genres=df_movies_genres[df_movies_genres['genres'].isin(top_genres)]
fig,ax=plt.subplots(figsize=(12,6))

##Boxplot
sns.boxplot(data=df_top_genres,y='genres',x='rating',palette='Set2',ax=ax,orient='h')
ax.set_title('Distribución de Ratings por Género (Top 10)\nStreamView Analytics',
            fontsize=14,fontweight='bold',pad=20)
ax.set_xlabel('Rating Promedio',fontsize=11)
ax.set_ylabel('Género',fontsize=11)
ax.grid(axis='x',alpha=0.3)

plt.tight_layout()
plt.savefig('boxplot_1_raring_por_genero.png',dpi=300,bbox_inches='tight')
plt.show()
```

```python
## STORYTELLING BOXPLOT 1
for genre in top_genres:
    genre_data=df_top_genres[df_top_genres['genres']==genre]['rating'].dropna()
    median_rating=genre_data.median()
    mean_rating=genre_data.mean()
    q1=genre_data.quantile(0.25)
    q3=genre_data.quantile(0.75)

    print(f"\n-------- {genre.upper()} --------")
    print(f"\n\nRating Mediano: {median_rating:.2f}")
    print(f"\nRating Promedio: {mean_rating:.2f}")
    print(f"\nRango Intercuartil (q1-q3): {q1:.2f}-{q3:.2f}\n\n")
```

# StoryTelling 1

## Análisis de confiabilidad

## Este análisis responde: 
- ¿Todos los géneros de Netflix tienen la misma calidad?
- ¿Cuáles generan más confianza en usuarios?

- Métrica usada: Mediana (50% de películas) + IQR (consistencia)
  
Medimos esto por la MEDIANA Y CONSISTENCIA (IQR)

### HALLAZGO 1 : Confiabilidad de Géneros en Netflix | Análisis de 16.000

Contexto:
Netflix tiene 16.000 películas. Cuando un uusario busca algo, ¿confía realmente en encontrar calidad? 
La respuesta depende del GÉNERO.

## Los Datos Revelan

### Género seguros (Rating cosistentemente alto):

##### DOCUMENTARY: 7.10/10 (RANGO 6.70-7.50)
- 75% de documentales tienen rating entre 6.70-7.50
- Variabilidad BAJA = usuarios saben qué esperar
- Comportamiento: Usuario clickea SIN leer sinopsis

##### DRAMA: 6.60/10 (rango 5.90-7.20)
- 50% de películas entre 6.00-7.10
- Caja estrecha = Consistencia
- Comportamiento: "Drama = calidad garantizada"

##### DRAMA + ROMANCE: 6.60/10 (rango 5.90-7.20)
- Hallazgo CLAVE: mezclar géneros SUBE confiabilidad
- Comparar: Romance puro (5.60) vs Drama+Romance (6.60)
- Interpretación: Drama "rescata" Romance de la mediocridad

### Géneros Mediocres (Rating bajo + variable):

#### HORROR: 5.60/10 (rango 5.00-6.13)
- Caja estrecha PERO baja
- Solo 25% de películas supera rating 6.13
- Comportamiento: Usuario desconfía

#### ROMANCE (solo): 5.60/10 (rango 4.31-6.88) <- CRÍTICO
- Rangp AMPLÍSIMO: varía 2.57 puntos
- Interpretación: 50% de películas Romance es "lotería"
- Variabilidad extrema = impredecible = churn
- Comportamiento: "¿será buena o mala? Mejor no arriesgo"
